LATIHAN 1

```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Jawaban : Jika menggunakan list Python biasa, [5000,7000,3000] * 0.9 akan menghasilkan error, karena list Python tidak dapat dikalikan langsung dengan bilangan desimal. Berbeda dengan NumPy array yang dapat melakukan operasi perkalian secara langsung ke seluruh elemen array

LATIHAN 2

```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Jawaban : Kolom yang memiliki data kosong (None) adalah kolom menu dan kolom terjual. Jika data tersebut langsung dianalisis tanpa dibersihkan, hasil analisis dapat menjadi kurang akurat karena terdapat data yang belum lengkap.  

LATIHAN 3

```python
df = pd.read_csv('data_kantin (1).csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Jawaban : Kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris total adalah kolom yang memiliki data kosong (missing value). Artinya, tidak semua baris pada kolom tersebut memiliki data atau nilai.

LATIHAN 4

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Jawaban : Kolom terjual diisi dengan 0 karena data yang kosong masih dapat diberi nilai yang sesuai untuk melanjutkan analisis. Sedangkan baris dengan menu kosong dihapus karena kolom menu diperlukan untuk mengetahui nama atau jenis menu yang dianalisis. Jika menu kosong, baris tersebut tidak dapat digunakan dengan baik dalam analisis berdasarkan menu

LATIHAN 5

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Jawaban : Setelah drop_duplicates(), jumlah baris dataset akan berkurang jika terdapat data yang duplikat. Jika tidak terdapat duplikat, jumlah baris tetap sama. Memastikan tipe data sudah benar penting agar data dapat diolah dan dihitung dengan tepat

LATIHAN 6

```python
laris = df[df['terjual'] > 20] # filtering memilih data menu yang terjual lebih dari 20.
urut = df.sort_values(by='terjual', ascending=False) # sorting mengurutkan data berdasarkan jumlah terjual.
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan menghitung total pendapatan setiap menu.
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi mengelompokkan data berdasarkan nama menu, kemudian menjumlahkan seluruh pendapatan dari masing-masing menu.
print(ringkasan)
```

JAWABAN : Berdasarkan hasil groupby, menu yang menghasilkan total_pendapatan tertinggi adalah Nasi Goreng, yaitu sebesar Rp3.924.000. Informasi ini dapat membantu kantin sekolah dalam menentukan menu yang perlu diprioritaskan, seperti menambah stok Nasi Goreng agar tidak kehabisan dan mempertahankan menu yang banyak menghasilkan pendapatan.
